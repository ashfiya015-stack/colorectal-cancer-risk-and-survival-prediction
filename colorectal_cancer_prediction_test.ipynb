# Colorectal Cancer Risk & Survival Prediction

## Project Overview

This project performs an exploratory data analysis of a colorectal cancer patient dataset.

The analysis focuses on understanding patterns and relationships among:

- Demographic characteristics
- Lifestyle and dietary factors
- Family and previous cancer history
- Cancer stage and tumor aggressiveness
- Screening and healthcare access
- Treatment-related factors
- Survival status
- Recurrence patterns

The goal is to generate meaningful, data-driven insights from the dataset rather than build a machine learning prediction model.

## 1. Project Objectives

The main objectives of this analysis are to:

1. Understand the structure and characteristics of the dataset.
2. Assess data quality, including missing values and duplicate records.
3. Explore demographic and clinical characteristics of patients.
4. Analyze lifestyle and potential risk-related factors.
5. Examine survival patterns across different patient groups.
6. Investigate the relationship between cancer stage, tumor aggressiveness, treatment, and survival.
7. Analyze recurrence patterns among patients.
8. Identify important patterns and relationships using statistical summaries and visualizations.
9. Present actionable insights and conclusions from the analysis.

> **Note:** This is an exploratory and descriptive data analysis project. Observed relationships should not be interpreted as causal or as medical advice.
